```python
import numpy as np

def generate_dummy_transect():
    # 1. Setup Time (10 minutes at 1Hz)
    t = np.arange(600) 
    
    # 2. Starting Point (The "Origin" near the front)
    lat_start, lon_start = 35.0, -98.0
    
    # 3. Create a Zigzag Path
    # Primary movement is East (lon increases)
    # Secondary movement is a North-South sine wave (the zigzag)
    lats = lat_start + (0.00001 * t) + (0.005 * np.sin(2 * np.pi * t / 200))
    lons = lon_start + (0.00015 * t)
    
    return lats, lons

# Generate the test arrays
lats, lons = generate_dummy_transect()
```

```python
test = BR_transform(lats, lons, frontal_crossing=[35.0, -98.0], bndry_angle=315)
```
